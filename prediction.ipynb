```python
import tensorflow as tf
```

```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import load_model
import pickle
```

```python
model = load_model('model.h5')

with open('lable_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file)

with open('onehot_encoder.pkl','rb') as file:
    onehot_encoder = pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
input_data = {
    "credit_score": 650,
    "country": "France",
    "gender": "Male",
    "age": 40,
    "tenure": 5,
    "balance": 60000.0,
    "products_number": 2,
    "credit_card": 1,
    "active_member": 1,
    "estimated_salary": 50000.0
}
```

```python
country_encoded = onehot_encoder.transform([[input_data['country']]]).toarray()
country_encoded_df = pd.DataFrame(country_encoded, columns=onehot_encoder.get_feature_names_out(['country']))
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
input_df['gender'] = label_encoder_gender.transform(input_df['gender'])
```

```python
input_df = pd.concat([input_df.drop(['country'],axis=1),country_encoded_df],axis=1)
```

```python
input_df
```

```python
input_scaled=scaler.transform(input_df)
```

```python
prediction=model.predict(input_scaled)
prediction
```

```python
if prediction[0][0] > 0.5:
    print("Customer is likely to churn")
else:
    print("Customer is not likely to churn")
```
